```python
# Calculate the Haversine distance between monitoring stations

def haversine_distance_km(
    lat1,
    lon1,
    lat2,
    lon2
):
    """
    Calculate the great-circle distance between two geographic
    coordinates using the Haversine formula.
    """

    earth_radius_km = 6371.0088

    lat1_rad = np.radians(lat1)
    lon1_rad = np.radians(lon1)
    lat2_rad = np.radians(lat2)
    lon2_rad = np.radians(lon2)

    delta_lat = lat2_rad - lat1_rad
    delta_lon = lon2_rad - lon1_rad

    a = (
        np.sin(delta_lat / 2) ** 2
        + np.cos(lat1_rad)
        * np.cos(lat2_rad)
        * np.sin(delta_lon / 2) ** 2
    )

    c = 2 * np.arctan2(
        np.sqrt(a),
        np.sqrt(1 - a)
    )

    return earth_radius_km * c

station_metadata["Station_Distance_km"] = (
    haversine_distance_km(
        station_metadata["Water_Latitude"],
        station_metadata["Water_Longitude"],
        station_metadata["Climate_Latitude"],
        station_metadata["Climate_Longitude"]
    )
)

station_metadata["Station_Distance_km"] = (
    station_metadata["Station_Distance_km"]
    .round(1)
)

display(
    station_metadata[
        [
            "Lake",
            "Water_Station_ID",
            "Water_Station_Name",
            "Climate_Station_ID",
            "Climate_Station_Name",
            "Station_Distance_km"
        ]
    ]
)
```

```python
import plotly.graph_objects as go

import warnings
warnings.filterwarnings(
    "ignore",
    category=UserWarning,
    module=r"kaleido\._sync_server"
)

fig = go.Figure()



# 1. Add connecting lines between paired monitoring stations

for _, row in station_metadata.dropna(
    subset=[
        "Water_Latitude",
        "Water_Longitude",
        "Climate_Latitude",
        "Climate_Longitude"
    ]
).iterrows():

    fig.add_trace(
        go.Scattergeo(
            lon=[
                row["Water_Longitude"],
                row["Climate_Longitude"]
            ],
            lat=[
                row["Water_Latitude"],
                row["Climate_Latitude"]
            ],
            mode="lines",
            line=dict(
                width=1,
                color="rgba(100,100,100,0.6)",
                dash="dot"
            ),
            hoverinfo="skip",
            showlegend=False
        )
    )



# 2. Create map labels for the water-level stations

station_metadata["Map_Label"] = [
    "Superior",
    "Huron",
    "Michigan",
    "Erie",
    "Ontario",
    "Woods",
    "Great Salt Lake",
    "Champlain",
    "Red Lake",
    "Tahoe"
]


water_text_position_map = {
    "Lake Superior": "bottom right",
    "Lake Huron": "top right",
    "Lake Michigan": "bottom left",
    "Lake Erie": "bottom center",
    "Lake Ontario": "top right",
    "Lake Woods": "top left",
    "Great Salt Lake": "top center",
    "Lake Champlain": "top right",
    "Red Lake": "bottom left",
    "Lake Tahoe": "top left"
}

station_metadata["Water_Text_Position"] = (
    station_metadata["Lake"]
    .map(water_text_position_map)
)


# 3. Add water-level monitoring stations

water_map_data = station_metadata.dropna(
    subset=[
        "Water_Latitude",
        "Water_Longitude"
    ]
).copy()


fig.add_trace(
    go.Scattergeo(
        lon=water_map_data["Water_Longitude"],
        lat=water_map_data["Water_Latitude"],

        mode="markers+text",

        marker=dict(
            size=7,
            symbol="circle",
            line=dict(width=0.8)
        ),

        text=water_map_data["Map_Label"],

        textposition=water_map_data["Water_Text_Position"],

        customdata=water_map_data[
            [
                "Lake",
                "Water_Station_ID",
                "Water_Station_Name"
            ]
        ],

        hovertemplate=(
            "<b>%{customdata[0]}</b><br>"
            "Water-level station: %{customdata[1]}<br>"
            "Station name: %{customdata[2]}<br>"
            "Latitude: %{lat:.4f}<br>"
            "Longitude: %{lon:.4f}"
            "<extra></extra>"
        ),

        name="Water-level station"
    )
)



# 4. Add climate monitoring stations

climate_map_data = station_metadata.dropna(
    subset=[
        "Climate_Latitude",
        "Climate_Longitude"
    ]
).copy()


fig.add_trace(
    go.Scattergeo(
        lon=climate_map_data["Climate_Longitude"],
        lat=climate_map_data["Climate_Latitude"],

        mode="markers",

       marker=dict(
           size=10,
           symbol="diamond-open",
           line=dict(width=1.8)
       ),

        customdata=climate_map_data[
            [
                "Lake",
                "Climate_Station_ID",
                "Climate_Station_Name",
                "Station_Distance_km"
            ]
        ],

        hovertemplate=(
            "<b>%{customdata[0]}</b><br>"
            "Climate station: %{customdata[1]}<br>"
            "Station name: %{customdata[2]}<br>"
            "Distance from water-level station: "
            "%{customdata[3]:.1f} km<br>"
            "Latitude: %{lat:.4f}<br>"
            "Longitude: %{lon:.4f}"
            "<extra></extra>"
        ),

        name="Climate station"
    )
)

# 5. Set the geographic extent and map appearance

fig.update_geos(
    projection_type="mercator",

    lonaxis=dict(
        range=[-125, -68]
    ),

    lataxis=dict(
        range=[37, 51]
    ),

    showland=True,
    landcolor="rgb(245, 245, 242)",

    showocean=True,
    oceancolor="rgb(225, 235, 245)",

    showlakes=True,
    lakecolor="rgb(225, 235, 245)",

    showcountries=True,
    countrycolor="rgba(80,80,80,0.8)",

    showsubunits=True,
    subunitcolor="rgba(130,130,130,0.5)",

    coastlinecolor="rgba(80,80,80,0.8)"
)


# 6. Final figure layout

fig.update_layout(
    width=1100,
    height=560,

    margin=dict(
        l=10,
        r=10,
        t=10,
        b=45
    ),

    legend=dict(
        orientation="h",
        x=0.5,
        xanchor="center",
        y=0.01,
        yanchor="top",
        font=dict(size=13)
    )
)

fig.show()
```
